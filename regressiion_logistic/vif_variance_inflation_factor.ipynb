```python
#VIF Calculation


import pandas as pd
import statsmodels.formula.api as sm

customers = pd.read_csv('datasets/customer_income_expenses.csv')
customers.head()

def vif_cal(x_vars):
    xvar_names = x_vars.columns
    for i in range(0,xvar_names.shape[0]):
        y = x_vars[xvar_names[i]]
        x = x_vars[xvar_names.drop(xvar_names[i])]
        rsq = sm.ols(formula='y ~ x', data=x_vars).fit().rsquared
        vif = round(1/(1 - rsq),2)
        print (xvar_names[i], " VIF =  ", vif)

vif_cal(x_vars=customers.drop('Monthly_Expenses', axis=1))
```

```python
# all four independent variable have VIF >5, indicating multicollinearity, but we cannot remove all

# we can drop the variable with highest VIF first and recalculate VIFs

vif_cal(x_vars=customers.drop(['Monthly_Expenses','Monthly_Income_in_Euro'], axis=1))
```

```python
# drop Number of personal loans and recalculate VIFs

vif_cal(x_vars=customers.drop(['Monthly_Expenses','Monthly_Income_in_Euro','Number_of_personal_loans'], axis=1))
```

```python
# only these two variables are left now with VIF <5, so we can go ahead with these two independent variables

model_final = sm.ols(formula='Monthly_Expenses ~ Number_of_Credit_cards + Monthly_Income_in_USD', data=customers)
fitted_final = model_final.fit()
fitted_final.summary()
```
