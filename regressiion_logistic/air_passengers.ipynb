```python
#Problem Statement: What will be the number of air passengers when the marketing cost = $4500K

import pandas as pd
import matplotlib.pyplot as plt
import statsmodels.formula.api as sm

air_passengers = pd.read_csv('air_passengers.csv')

air_passengers.head()

air_passengers.info()

air_passengers.describe()
```

```python
# y axis ~ x axis
model1 = sm.ols(formula='Passengers_count ~ marketing_cost', data=air_passengers)

fitted1 = model1.fit()
fitted1.summary()
```

```python

#no of passengers were expecting when marketing cost is 4500K
new_data = pd.DataFrame({'marketing_cost':[4500]})
print(fitted1.predict(new_data))
```

```python
# predictions

air_passengers['passengers_count_predicted'] = round(fitted1.predict(air_passengers))

keep_cols = ["marketing_cost", "Passengers_count", "passengers_count_predicted"]

air_passengers[keep_cols]

air_passengers.columns
```

```python
# vs customer_ratings
model2 = sm.ols(formula='Passengers_count ~ customer_ratings', data=air_passengers)

fitted2 = model2.fit()
fitted2.summary()
```

```python
# inference: coffecient of R square = 0.102 when modeled with customer_ratings
# vs 0.761 when modeled with marketing_cost
# 76.1 % variation in number of passengers can be explained by marketing_cost
# only 10.2 % variation in number of passengers can be explained by customer_ratings
```

```python
#multiple linear regression
# air_passengers.columns
model3 = sm.ols(formula='Passengers_count ~ marketing_cost+customer_ratings +percent_delayed_flights + number_of_trips + poor_weather_index +percent_male_customers + Holiday_week + percent_female_customers'
                        , data=air_passengers)
fitted3 = model3.fit()
fitted3.summary()

#coeff of r squar is 0.911 which mean 91.1 % variation in number of passengers can be explained by all the independent variables together
```

```python
# multicollinearity check - having too many independent variables can be a curse

# VIF calculation function

def vif_cal(x_vars):
    xvar_names = x_vars.columns
    for i in range(0,xvar_names.shape[0]):
        y = x_vars[xvar_names[i]]
        x = x_vars[xvar_names.drop(xvar_names[i])]
        rsq = sm.ols(formula='y ~ x', data=x_vars).fit().rsquared
        vif = round(1/(1 - rsq),2)
        print (xvar_names[i], " VIF =  ", vif)

vif_cal(x_vars=air_passengers.drop(['Passengers_count','passengers_count_predicted'], axis=1))
```

```python
# Dropping variables with high VIF one by one and recalculating VIFs

vif_cal(x_vars=air_passengers.drop(['Passengers_count','passengers_count_predicted','percent_male_customers'], axis=1))
```

```python
# Dropping variables with high VIF one by one and recalculating VIFs

vif_cal(x_vars=air_passengers.drop(['Passengers_count','passengers_count_predicted','percent_male_customers','percent_delayed_flights'], axis=1))
```

```python
# model after removing variables with high VIF
model_final = sm.ols(formula='Passengers_count ~ marketing_cost + customer_ratings + number_of_trips + poor_weather_index + Holiday_week + percent_female_customers',data=air_passengers)
fitted_final = model_final.fit()
fitted_final.summary()

#r square dropped from 0.911 to 0.904 after removing two variables with high VIF, so we can go ahead with this model
```
