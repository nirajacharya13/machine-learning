```python
import pandas as pd
import matplotlib.pyplot as plt
emp_profile = pd.read_csv('employee_profile.csv')
```

```python
print(emp_profile.head())

print(emp_profile.info())

print (emp_profile.describe())
```

```python
## Drawing the Scatter Plot between Monthly Income And Expenses

import matplotlib.pyplot as plt

plt.scatter(emp_profile['Monthly_Income'], emp_profile['Monthly_Expenses'])
plt.title('Monthly Income vs Monthly Expense')      
plt.xlabel('Monthly Income')
plt.ylabel('Monthly Expense')
plt.show()
```

```python
plt.scatter(emp_profile['Monthly_Income'], emp_profile['Time_Spent_Reading_Books'])
plt.title('Monthly Income vs Time Spent Reading Books')      
plt.xlabel('Monthly Income')
plt.ylabel('Time Spent Reading Books')
plt.show()
```
