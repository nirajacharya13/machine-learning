```python
import pandas as pd
import statsmodels.formula.api as sm
import matplotlib.pyplot as plt

customers = pd.read_csv('customer_income_expenses.csv')
customers.head()
```

```python
model1 = sm.ols(formula='Monthly_Expenses ~ Monthly_Income_in_USD + Number_of_Credit_cards + Number_of_personal_loans + Monthly_Income_in_Euro', data=customers)

fitted1 = model1.fit()
fitted1.summary()

# incom in USD and Euro should have same correlation. but the intercepts are 7.5 and -7.6 resp, which
# indicates multicollinearity issue. so we can remove one of these variables

# removing Monthly_Income_in_Euro

```

```python
model2 = sm.ols(formula='Monthly_Expenses ~ Number_of_Credit_cards + Number_of_personal_loans + Monthly_Income_in_Euro', data=customers)

fitted2 = model2.fit()
fitted2.summary()

# coeff of r square did not change after removing one variable, i.e. income in USD, so we can go ahead with this model
```
