```python
import pandas as pd
import statsmodels.formula.api as sm
import matplotlib.pyplot as plt

product_Sales = pd.read_csv('datasets\Product_sales.csv')
product_Sales.describe()

# product_Sale.value_counts
```

```python
model = sm.ols(formula='Bought ~ Income', data=product_Sales)
fitted = model.fit()
fitted.summary()
```

```python
# drawing the scatter plot with regression line
import matplotlib.pyplot as plt

plt.scatter(product_Sales['Income'], product_Sales['Bought'], color='blue')
plt.title('Income vs Bought')
plt.xlabel('Income')
plt.ylabel('Bought')
plt.show()
```

```python
pred_values = fitted.predict(product_Sales['Income'])
plt.scatter(product_Sales['Income'], product_Sales['Bought'], color='blue')
plt.plot(product_Sales['Income'], pred_values, color='red')
plt.title('Income vs Bought')
plt.xlabel('Income')
plt.ylabel('Bought')
plt.show()
```

```python
import statsmodels.api as sm
logit_model = sm.Logit(product_Sales['Bought'], product_Sales[['Income']])

logit_model1 = sm.Logit(product_Sales['Bought'], sm.add_constant(product_Sales[['Income']]))
logit_model1_fit = logit_model1.fit()
logit_model1_fit.summary()
```
